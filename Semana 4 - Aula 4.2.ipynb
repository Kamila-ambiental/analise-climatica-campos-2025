## 1. OBJETIVO
O objetivo deste projeto é coletar, organizar e analisar dados climáticos históricos de Campos dos Goytacazes/RJ referentes ao ano de 2025.
A análise busca compreender o comportamento das condições climáticas ao longo do período, com foco nas variações de temperatura, precipitação e velocidade do vento.
A partir dos dados coletados, pretende-se identificar períodos mais quentes, mais frios e mais chuvosos, além de aplicar técnicas básicas de tratamento, análise exploratória e visualização de dados.

## 3. METODOLOGIA
### 3.1 Coleta

A coleta dos dados foi realizada em Python por meio de uma requisição à Open-Meteo Historical Weather API.
Foram utilizadas as bibliotecas requests, pandas e matplotlib.
A biblioteca requests foi utilizada para realizar a requisição à API, o pandas para organizar, tratar e analisar os dados e o matplotlib para a elaboração das visualizações gráficas.
As variáveis selecionadas para o estudo foram:

- Data;
- Temperatura média diária;
- Temperatura máxima diária;
- Temperatura mínima diária;
- Precipitação diária;
- Velocidade máxima diária do vento.

Posteriormente, também foi criada uma variável referente ao mês para possibilitar as análises mensais.

### 3.2 Limpeza

Após a coleta, os dados retornados pela API foram convertidos para um DataFrame utilizando a biblioteca pandas.
Na etapa de tratamento foram realizados os seguintes procedimentos:

- organização dos dados em formato tabular;
- renomeação das colunas para facilitar a interpretação;
- conversão da coluna de data para o formato datetime;
- verificação da existência de valores ausentes;
- verificação da existência de registros duplicados;
- criação da variável correspondente ao mês de cada registro.

Esses procedimentos permitiram preparar o conjunto de dados para as etapas seguintes da análise.

### 3.3 Processamento

Após a limpeza e organização, foram realizadas análises estatísticas para compreender o comportamento das variáveis climáticas durante o ano de 2025.

Foram calculados:

- temperatura média do período;
- maior temperatura registrada;
- menor temperatura registrada;
- precipitação total acumulada;
- temperatura média mensal;
- precipitação acumulada mensal;
- mês com maior temperatura média;
- mês com menor temperatura média;
- mês com maior precipitação.

Também foram produzidos gráficos para auxiliar na interpretação dos resultados.

## 5. ANÁLISE EXPLORATÓRIA
### 5.1 Estatísticas Descritivas

A análise exploratória permitiu avaliar o comportamento das condições climáticas de Campos dos Goytacazes/RJ durante o ano de 2025.

Os principais resultados obtidos foram:

- Maior temperatura registrada: 40,1 °C

## 6. INSIGHTS
1. *Variação das temperaturas* - A temperatura média observada durante o ano de 2025 foi de 24,21 °C. Os valores extremos variaram entre uma temperatura mínima de 12,2 °C e uma temperatura máxima de 40,1 °C, demonstrando uma variação significativa ao longo do período analisado.
2. *Período mais quente* - Fevereiro apresentou a maior temperatura média mensal do período analisado. Além disso, foram registrados episódios de temperaturas máximas chegando a 40,1 °C.
3. *Período mais frio* - Julho apresentou a menor temperatura média mensal de 2025. Entretanto, o menor valor diário de temperatura mínima não ocorreu em julho. A menor temperatura registrada no conjunto de dados foi de 12,2 °C em 28 de setembro de 2025. Esse resultado demonstra a diferença entre analisar a média mensal das temperaturas e analisar o menor valor diário registrado durante o ano.
4.  *Precipitação* - A precipitação total acumulada durante o período analisado foi de 861,3 mm. Entre os meses avaliados, dezembro apresentou o maior volume acumulado de precipitação.
5.  *Comportamento ao longo do ano* - Os dados permitiram observar diferenças no comportamento das temperaturas e da precipitação ao longo de 2025. As temperaturas médias foram mais elevadas no início do ano, apresentaram redução durante os meses intermediários e voltaram a aumentar posteriormente. A precipitação também apresentou variações mensais, com maior volume acumulado em dezembro.

## 7. LIMITAÇÕES
- A análise considera apenas o ano de 2025, não sendo suficiente para identificar tendências climáticas de longo prazo.
- Os dados representam as coordenadas selecionadas de Campos dos Goytacazes/RJ e podem não refletir variações climáticas de todo o município.
